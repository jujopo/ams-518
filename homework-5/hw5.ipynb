### Problem 1   

Suppose we know that the error in the risk quadrangle is the mean squared error, defined by
$$\mathcal{E}(X) = \lambda \|X\|_2^2 = E[X^2]$$
Use the relations in Diagram 3 to derive all remaining elements in the risk quadrangle (Regret, Statistic, Deviation, Error). Follow these steps:

1.  Derive the **regret** in the risk quadrangle with the relation
    $$\mathcal{V}(X) = \mathcal{E}(X) + E[X]$$
2.  Derive the **risk** in the risk quadrangle with the relation
    $$\mathcal{R}(X) = \min_C \{C + \mathcal{V}(X-C)\}$$
3.  Derive the **statistic** in the risk quadrangle with the relations
    $$S(X) = \arg\min_C \mathcal{E}(X-C), \quad S(X) = \arg\min_C \{C + \mathcal{V}(X-C)\}$$
4.  Derive the **deviation** in the risk quadrangle with the relation
    $$\mathcal{D}(X) = \mathcal{R}(X) - E[X]$$

***

### Solution

We are given the error measure $\mathcal{E}(X) =\lambda \|X\|_2^2 = E[X^2]$. We will now derive the other three components of the risk quadrangle.

1. Derivation of Regret $\mathcal{V}(X)$

We use the given relation $\mathcal{V}(X) = \mathcal{E}(X) + E[X]$.

Substituting the definition of $\mathcal{E}(X)$:
$$\mathcal{V}(X) = E[X^2] + E[X]$$

2. Derivation of Risk $\mathcal{R}(X)$

The risk is defined as $\mathcal{R}(X) = \min_{C} \{C + \mathcal{V}(X - C)\}$. First, let's express $\mathcal{V}(X - C)$ using the formula for regret we just found.

$$\mathcal{V}(X - C) = E[(X-C)^2] + E[X-C]$$

Expanding the terms:$$E[(X-C)^2] = E[X^2 - 2XC + C^2] = E[X^2] - 2CE[X] + C^2$$

$$E[X-C] = E[X] - C$$So,$$\mathcal{V}(X - C) = (E[X^2] - 2CE[X] + C^2) + (E[X] - C)$$

Now, substituting this into the expression for $\mathcal{R}(X)$:

$$\mathcal{R}(X) = \min_{C} \{C + (E[X^2] - 2CE[X] + C^2) + (E[X] - C)\}$$
$$\mathcal{R}(X) = \min_{C} \{C^2 - 2CE[X] + E[X^2] + E[X]\}$$

To find the value of $C$ that minimizes this expression, we take the derivative with respect to $C$ and set it to zero. 

Let $f(C) = C^2 - 2CE[X] + E[X^2] + E[X]$.

$$\frac{df}{dC} = 2C - 2E[X] = 0$$
$$\implies C = E[X]$$

Now we plug this optimal value of $C$ back into the expression to find the minimum value, which is $\mathcal{R}(X)$:

$$\mathcal{R}(X) = (E[X])^2 - 2(E[X])E[X] + E[X^2] + E[X]$$
$$\mathcal{R}(X) = (E[X])^2 - 2(E[X])^2 + E[X^2] + E[X]$$
$$\mathcal{R}(X) = E[X^2] - (E[X])^2 + E[X]$$

Since the variance is defined as $\text{Var}(X) = E[X^2] - (E[X])^2$, the **risk** is:

$$\mathcal{R}(X) = \text{Var}(X) + E[X]$$

3. Derivation of Statistic $S(X)$

The statistic $S(X)$ is the value of $C$ that achieves the minimum in the risk calculation. We can find it using either of the two given relations.

**Method 1:** Using $S(X) = \arg\min_C \{C + \mathcal{V}(X-C)\}$.
From our work in Step 2, we found that the expression was minimized when $C = E[X]$, hence:
$$S(X) = E[X]$$

**Method 2:** Using $S(X) = \arg\min_C \mathcal{E}(X - C)$.
$$\mathcal{E}(X-C) = E[(X-C)^2]$$

Let's minimize this with respect to $C$. Let $g(C) = E[(X-C)^2] = E[X^2] - 2CE[X] + C^2$.

$$\frac{dg}{dC} = -2E[X] + 2C = 0$$
$$\implies C = E[X]$$

Both methods confirm that the optimal **statistic** is the expected value

$$S(X) = E[X]$$

4. Derivation of Deviation $\mathcal{D}(X)$

The deviation is defined by the relation $\mathcal{D}(X) = \mathcal{R}(X) - E[X]$.

Using our result for the risk $\mathcal{R}(X)$ from Step 2

$$\mathcal{D}(X) = (\text{Var}(X) + E[X]) - E[X]$$

This simplifies to:

$$\mathcal{D}(X) = \text{Var}(X)$$

The **deviation** is the variance of $X$.
